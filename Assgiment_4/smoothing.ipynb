```python
import pickle, os
from collections import defaultdict



```

```python
MODEL_DIR = "ngram_model"
```

```python
def add_one_smoothing(ngram_size: int):
    # Load counts
    with open(os.path.join(MODEL_DIR, f"final_{ngram_size}gram_counts.pkl"), "rb") as f:
        ngram_counts = pickle.load(f)
    with open(os.path.join(MODEL_DIR, f"final_{ngram_size}gram_context_counts.pkl"), "rb") as f:
        context_counts = pickle.load(f)

    # Vocabulary size (number of unique tokens)
    vocab = set([ngram[-1] for ngram in ngram_counts.keys()])
    V = len(vocab)

    smoothed_probs = {}
    for ngram, count in ngram_counts.items():
        context = ngram[:-1] if ngram_size > 1 else ()
        smoothed_probs[ngram] = (count + 1) / (context_counts[context] + V)

    # Save smoothed probabilities
    with open(os.path.join(MODEL_DIR, f"smoothed_addone_{ngram_size}gram_probs.pkl"), "wb") as f:
        pickle.dump(smoothed_probs, f)

    print(f"[Add-One Smoothing] {ngram_size}-gram done. Saved {len(smoothed_probs)} entries.")

if __name__ == "__main__":
    for n in [1,2,3,4]:
        add_one_smoothing(n)
```

```python

K = 0.5   # you can change k (e.g., 0.1, 0.5, 1)

def add_k_smoothing(ngram_size: int, k=K):
    # Load counts
    with open(os.path.join(MODEL_DIR, f"final_{ngram_size}gram_counts.pkl"), "rb") as f:
        ngram_counts = pickle.load(f)
    with open(os.path.join(MODEL_DIR, f"final_{ngram_size}gram_context_counts.pkl"), "rb") as f:
        context_counts = pickle.load(f)

    # Vocabulary size
    vocab = set([ngram[-1] for ngram in ngram_counts.keys()])
    V = len(vocab)

    smoothed_probs = {}
    for ngram, count in ngram_counts.items():
        context = ngram[:-1] if ngram_size > 1 else ()
        smoothed_probs[ngram] = (count + k) / (context_counts[context] + k*V)

    # Save smoothed probabilities
    with open(os.path.join(MODEL_DIR, f"smoothed_addk_{ngram_size}gram_probs.pkl"), "wb") as f:
        pickle.dump(smoothed_probs, f)

    print(f"[Add-K Smoothing] {ngram_size}-gram done with k={k}. Saved {len(smoothed_probs)} entries.")

if __name__ == "__main__":
    for n in [1,2,3,4]:
        add_k_smoothing(n, k=0.5)
```

```python
def token_type_smoothing(ngram_size: int):
    # Load counts
    with open(os.path.join(MODEL_DIR, f"final_{ngram_size}gram_counts.pkl"), "rb") as f:
        ngram_counts = pickle.load(f)
    with open(os.path.join(MODEL_DIR, f"final_{ngram_size}gram_context_counts.pkl"), "rb") as f:
        context_counts = pickle.load(f)

    # Token types (unique words in model)
    token_types = set([ngram[-1] for ngram in ngram_counts.keys()])
    T = len(token_types)

    smoothed_scores = {}
    for ngram, count in ngram_counts.items():
        context = ngram[:-1] if ngram_size > 1 else ()
        smoothed_scores[ngram] = (count + T) / (context_counts[context] + T)

    # Save smoothed scores (not normalized probs)
    with open(os.path.join(MODEL_DIR, f"smoothed_tokentype_{ngram_size}gram.pkl"), "wb") as f:
        pickle.dump(smoothed_scores, f)

    print(f"[Token-Type Smoothing] {ngram_size}-gram done. Saved {len(smoothed_scores)} entries.")

if __name__ == "__main__":
    for n in [1,2,3,4]:
        token_type_smoothing(n)
```
