```python
# from automathon import DFA
from graphviz import Digraph
import string
```

```python
# File paths
input_file = "brown_nouns.txt"
result_file="Processed_brown_nouns.txt"
accepted_file = "brown_nouns_accepted.txt"
rejected_file = "brown_nouns_rejected.txt"
stats_file = "stats.txt"
```

```python
# Define DFA states and symbols
states = {'q0', 'q1', 'q_reject'}
input_symbols = set([chr(c) for c in range(ord('a'), ord('z')+1)] + ['other'])

start_state = 'q0'
accept_states = {'q1'}
```

```python

transition_table={}

for c in input_symbols:
    transition_table[('q0',c)]='q1' if c!='other' else 'q_reject' 

for c in input_symbols:
    transition_table[('q1',c)]='q1' if c!='other' else 'q_reject' 

for c in input_symbols:
    transition_table[('q_reject',c)]='q_reject'     
```

```python
def classify_word(word):
    curr_state='q0'

    for c in word:
        if c in input_symbols:
            curr_state=transition_table[(curr_state,c)]
        else: 
            return False    
    return curr_state in accept_states
```

```python


def simplify_label(symbols):
    """Convert full a-z list to a short 'a–z' string if possible."""
    if set(symbols) == set(string.ascii_lowercase):
        return 'a–z'
    return ', '.join(sorted(symbols))

def visualize_dfa_clean(states, input_symbols, transition_table, start_state, accept_states, filename='clean_dfa'):
    dot = Digraph(format='png')
    dot.attr(rankdir='LR')  # Left to right layout

    # Draw states
    for state in states:
        if state in accept_states:
            dot.node(state, shape='doublecircle')
        elif 'reject' in state:
            dot.node(state, shape='circle', style='filled', fillcolor='lightcoral')
        else:
            dot.node(state, shape='circle')

    # Fake invisible start arrow
    dot.node('', shape='none')
    dot.edge('', start_state)

    # Collect edges: (src, dst) → list of input symbols
    edge_labels = {}
    for (src, symbol), dst in transition_table.items():
        key = (src, dst)
        edge_labels.setdefault(key, []).append(symbol)

    # Draw transitions with grouped labels
    for (src, dst), symbols in edge_labels.items():
        label = simplify_label(symbols)
        dot.edge(src, dst, label=label)

    # Render and open
    dot.render(filename, view=True)



def visualize_dfa_individual_arrows(states, input_symbols, transition_table, start_state, accept_states, filename='dfa_individual_arrows'):
    dot = Digraph(format='png')
    dot.attr(rankdir='LR')  # Left to right layout

    # Draw states
    for state in states:
        if state in accept_states:
            dot.node(state, shape='doublecircle')
        elif 'reject' in state:
            dot.node(state, shape='circle', style='filled', fillcolor='lightcoral')
        else:
            dot.node(state, shape='circle')

    # Start arrow
    dot.node('', shape='none')
    dot.edge('', start_state)

    # Draw individual transitions (no grouping)
    for (src, symbol), dst in transition_table.items():
        dot.edge(src, dst, label=symbol)

    # Render
    dot.render(filename, view=True)


# Visualize with individual arrows
visualize_dfa_individual_arrows(states, input_symbols, transition_table, start_state, accept_states)


# Visualize
# visualize_dfa_clean(states, input_symbols, transition_table, start_state, accept_states)
```

```python



accepted_words = []
not_accepted_words = []

# Read input words
with open(input_file, "r", encoding="utf-8") as f:
    words = [line.strip() for line in f if line.strip()]


    with open(result_file, "w", encoding="utf-8") as f:
        
        for word in words:
            if classify_word(word):
                accepted_words.append(word)
                f.write(f"{word} : Accepted\n")
            else:
                not_accepted_words.append(word)
                f.write(f"{word} : Rejected\n")

# Save accepted
with open(accepted_file, "w", encoding="utf-8") as f:
    f.write("\n".join(accepted_words))

# Save rejected
with open(rejected_file, "w", encoding="utf-8") as f:
    f.write("\n".join(not_accepted_words))

# Save stats
with open(stats_file, "w", encoding="utf-8") as f:
    f.write(f"Accepted words: {len(accepted_words)}\n")
    f.write(f"NOt Accepted words: {len(not_accepted_words)}\n")

print(" DFA processing complete!")

print(f"Accepted: {len(accepted_words)}, Rejected: {len(not_accepted_words)}")
```
